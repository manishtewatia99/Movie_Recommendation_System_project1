```python
import re
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.neural_network import MLPRegressor
from sklearn.metrics.pairwise import cosine_similarity

RANDOM_STATE = 42

def load_data(path):
    return pd.read_csv(path)

def clean_text(text):
    text = str(text).lower()
    text = re.sub(r"[^a-z0-9\s]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

def build_tags(df):
    df = df.copy()
    df["tags"] = (
        df["genres"].fillna("") + " " +
        df["cast"].fillna("") + " " +
        df["director"].fillna("") + " " +
        df["keywords"].fillna("") + " " +
        df["overview"].fillna("")
    )
    df["tags"] = df["tags"].apply(clean_text)
    return df

def vectorize_tags(df):
    vectorizer = TfidfVectorizer(max_features=3000, stop_words="english")
    tfidf_matrix = vectorizer.fit_transform(df["tags"]).toarray()
    return tfidf_matrix, vectorizer

def train_autoencoder(X):
    model = MLPRegressor(
        hidden_layer_sizes=(256, 64, 256),
        activation="relu",
        solver="adam",
        max_iter=1000,
        random_state=RANDOM_STATE
    )
    model.fit(X, X)
    return model

def relu(x):
    return np.maximum(0, x)

def extract_embeddings(model, X):
    hidden = relu(X @ model.coefs_[0] + model.intercepts_[0])
    embedding = relu(hidden @ model.coefs_[1] + model.intercepts_[1])
    return embedding

def recommend(title, df, embeddings, top_n=5):
    matches = df[df["title"].str.lower() == title.lower()]

    if matches.empty:
        matches = df[df["title"].str.lower().str.contains(title.lower(), na=False)]
        if matches.empty:
            return None, None

    idx = matches.index[0]
    similarity = cosine_similarity([embeddings[idx]], embeddings)[0]
    indices = np.argsort(similarity)[::-1]
    indices = [i for i in indices if i != idx][:top_n]

    result = pd.DataFrame({
        "Movie": df.loc[indices, "title"].values,
        "Genres": df.loc[indices, "genres"].values,
        "Similarity": np.round(similarity[indices], 4)
    })

    return df.loc[idx, "title"], result

df = load_data(r"C:\\Users\\Lenovo\\Downloads\\movie.csv")
df = build_tags(df)

tfidf_matrix, vectorizer = vectorize_tags(df)
autoencoder = train_autoencoder(tfidf_matrix)
embeddings = extract_embeddings(autoencoder, tfidf_matrix)

query = input("spider-Man: Brand New Day ")

movie, recommendations = recommend(query, df, embeddings)

if movie is None:
    print("Movie not found.")
else:
    print("\nSelected Movie:", movie)
    print(recommendations.to_string(index=False))
```
